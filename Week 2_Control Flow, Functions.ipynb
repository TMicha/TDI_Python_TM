Task 1: Order Approval Logic
Write a function called approval_message(order_value) that returns:
•
"Needs manager approval" if the value is above 100,000
•
"Needs supervisor approval" if it is above 50,000
•
"Auto-approved" otherwise
Test it with an order value of 72,500.

```python
def approval_message(order_value):
    if order_value > 100000:
       return "Needs Manager Approval"
    elif order_value > 50000:
        return "Needs Supervisor Approval"
    else:
         return "Auto-approved"

# testing the approval message function
print(approval_message(72500))
```

Task 3: Supplier Status Function
Create a function check_supplier_status(supplier_name, on_time_rate) that returns the supplier's name with a status:
•
"Reliable" if the on-time rate is 0.90 or higher
•
"Monitor" if it is 0.80 or higher
•
"At Risk" otherwise
Test it with at least 3 suppliers, for example ("SupplierA", 0.94), ("SupplierB", 0.86) and ("SupplierE", 0.74).

```python
def check_supplier_status(supplier_name, on_time_rate):
    if on_time_rate >= 0.90:
       return supplier_name + " - Reliable"
    elif on_time_rate >= 0.80:
       return supplier_name + " - Monitor"
    else:
       return supplier_name + " - At Risk"
    
```

Challenge 1: Write a function that takes an order value and a supplier's on-time rate, and returns a different approval level based on both. For example, reject orders above 100,000 from suppliers below 0.80.

```python
def order_supplier_approval(order_value, on_time_rate):

    if order_value > 100000 and on_time_rate < 0.80:
        return "Rejected"

    elif order_value > 100000 and on_time_rate >= 0.80:
        return "Needs manager approval"

    elif order_value > 50000 and on_time_rate < 0.80:
        return "Needs supervisor approval"

    elif order_value > 50000 and on_time_rate >= 0.80:
        return "Needs supervisor approval"

    else:
        return "Auto-approved"


# Test the supplier's on-time rate approval function
print(order_supplier_approval(150000, 0.74))
print(order_supplier_approval(150000, 0.94))
print(order_supplier_approval(72500, 0.86))
print(order_supplier_approval(40000, 0.94))
```

Challenge 2: Find all orders from SupplierA, however the name is written. Hint: name.lower().replace(" ", "") turns "Supplier A" into "suppliera". Count them and add up their total cost.

```python
# Initialise variables
supplier_count = 0
total_cost = 0

# Loop through all orders
for row in rows:

    # Standardise supplier name
    name = row["supplier_name"].lower().replace(" ", "")

    # Check if supplier is SupplierA
    if name == "suppliera":

        supplier_count += 1
        total_cost += float(row["total_cost"])

# Print results
print("Number of SupplierA orders:", supplier_count)
print("Total cost of SupplierA orders:", total_cost)
```

Challenge 3: Loop through the dataset and find rows with empty values (row[column] == ""). Which columns have gaps, and how many?

```python
# Create an empty dictionary to store missing value counts
missing_counts = {}

# Loop through each column
for column in rows[0].keys():

    missing_count = 0

    # Loop through each row
    for row in rows:

        if row[column] == "":
            missing_count += 1

    # Store the count for each column
    if missing_count > 0:
        missing_counts[column] = missing_count


# Print columns with missing values
print("Columns with missing values:")

for column, count in missing_counts.items():
    print(column, ":", count)
```
